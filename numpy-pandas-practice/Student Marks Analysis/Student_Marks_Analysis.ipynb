```python
import numpy as np
import pandas as pd
```

```python
data =pd.read_csv("/content/Studentdata.csv")
df =pd.DataFrame(data)
print(df.head())
print(df.info())
```

```python
mark = df[["MathScore","ScienceScore","EnglishScore"]]
total= np.sum(mark,axis=1)
average = total/3
df["total"] = total
df["Average"] = average
print(df.head())
```

```python
ScienceScore_firstclass = df[(df["ScienceScore"] >= 80) & (df["ScienceScore"] <= 100)]
print(ScienceScore_firstclass[["Name","ScienceScore"]])
```
